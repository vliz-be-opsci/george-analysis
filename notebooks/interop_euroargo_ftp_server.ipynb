### Understanding the file server structure

Navigating through the directory structure that contains various files and folders. 

```python
# Try to connect and list files
try:
    ftp = ftplib.FTP(ftp_url)
    ftp.login()
    ftp.cwd(ftp_path)
    files = ftp.nlst()
    ftp.quit()
    print(f"✅ Connected to {ftp_url}/{ftp_path}")
    print(f"📂 Directory listing :\n")
    for file in files:
        print(f" - {file}")
except Exception as e:
    print(f"❌ FTP connection failed: {e}")
```

```python
def list_ftp_directory(ftp_url, ftp_path):
    try:
        ftp = ftplib.FTP(ftp_url)
        ftp.login()
        ftp.cwd(ftp_path)
        files = ftp.nlst()
        ftp.quit()
        print(f"✅ Connected to {ftp_url}/{ftp_path}")
        print(f"📂 Directory listing :\n")
        for file in files:
            print(f" - {file}")
    except Exception as e:
        print(f"❌ FTP connection failed: {e}")

# Get content of list of .json files
def download_json_files(ftp, files):
    for file in files:
        if file.endswith('.json'):
            os.mkdir("files_euroargo_ftp") if not os.path.exists("files_euroargo_ftp") else None
            local_filename = os.path.join(os.getcwd(), "files_euroargo_ftp", file)
            with open(local_filename, 'wb') as local_file:
                ftp.retrbinary(f'RETR {file}', local_file.write)
            print(f"✅ Downloaded {file} to {local_filename}")
        else:
            print(f"⚠️ Skipped {file}, not a .json file")

# Read and print the content of the downloaded JSON file
def read_json_file(file_path):
    with open(file_path, 'r') as f:
        data = json.load(f)
    return data
```

```python
# Exploring a specific subdirectory
ftp_path = "ifremer/glider/v2/agathe/agathe_20220115"
list_ftp_directory(ftp_url, ftp_path)

# Downloading JSON files from the specified subdirectory
try:
    ftp = ftplib.FTP(ftp_url)
    ftp.login()
    ftp.cwd(ftp_path)
    files = ftp.nlst()
    download_json_files(ftp, files)
    ftp.quit()
except Exception as e:
    print(f"❌ Failed to download JSON files: {e}")
```

```python
# Check if the filenames follow any standardized or machine-readable patterns (e.g., ISO 8601 for dates)
import re

date_pattern = re.compile(r'\d{4}[-_]\d{2}[-_]\d{2}')

matching = [f for f in files if date_pattern.search(f)]
print(f"📅 Files matching ISO 8601-like date patterns: {len(matching)}")
print(f"📝 Sample: {matching[:3]}")
```

```python
# Check available file formats
extensions = [os.path.splitext(f)[1].lower() for f in files if '.' in f]
from collections import Counter
format_counts = Counter(extensions)

print("📦 File format distribution:")
for ext, count in format_counts.items():
    print(f"  - {ext if ext else '[no extension]'}: {count}")
```

```python
# list the files with .txt extension
txt_files = [f for f in files if f.endswith('.txt')]
txt_files
```

```python
# Get content of listed .txt files
def get_txt_content(ftp, filename):
    try:
        with open(filename, 'wb') as local_file:
            ftp.retrbinary(f'RETR {filename}', local_file.write)
        print(f"✅ Downloaded {filename}")
    except Exception as e:
        print(f"❌ Failed to download {filename}: {e}")

# Download .txt files
if txt_files:
    try:
        ftp = ftplib.FTP(ftp_url)
        ftp.login()
        ftp.cwd(ftp_path)
        
        for txt_file in txt_files:
            get_txt_content(ftp, txt_file)
        
        ftp.quit()
    except Exception as e:
        print(f"❌ FTP connection failed during download: {e}") 
```

content of .txt files same as the ones offered via the HTTPS file server... --> so same comments apply as for the HTTPS file server in context of semantic interoperability.
