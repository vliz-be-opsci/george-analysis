#### Standards compliance
*Involves checking compliance to HTTP standard, JSON schema, JSON API spec*

```python
# various endpoints
urls = [
    'https://sextant.ifremer.fr/examind/WS/sts/coriolis/v1.1',                  #base
    'https://sextant.ifremer.fr/examind/WS/sts/coriolis/v1.1/Things',
    'https://sextant.ifremer.fr/examind/WS/sts/coriolis/v1.1/Things(13108)',
    'https://sextant.ifremer.fr/examind/WS/sts/coriolis/v1.1/Things()',         #faulty endpoint
    'https://sextant.ifremer.fr/examind/WS/sts/coriolis/v1.1/Things(1318)'      #faulty endpoint - ID typo
]

# headers of interest
important_headers = [
    "Content-Type",
    "Content-Length",
    "Content-Encoding",
    "Transfer-Encoding",
    "Vary",
    "Accept",
    "Accept-Encoding",
    "Accept-Language",
    "Server",
    "Date",
    "Strict-Transport-Security"
]

for url in urls:
    print("="*100)
    print(f"🔗 URL: {url}")
    try:
        r = requests.get(url, timeout=10)
        print(f"Status code: {r.status_code}\n")

        # filter only important headers
        headers_filtered = {k: v for k, v in r.headers.items() if k in important_headers}
        
        print("📌 Relevant headers:")
        if headers_filtered:
            for k, v in headers_filtered.items():
                print(f"   {k}: {v}")
        else:
            print("   (none of the selected headers present)")
        
        # try parsing response
        print("\n📌 Response preview:")
        try:
            data = r.json()
            if isinstance(data, dict):
                print("   JSON object with keys:", list(data.keys()))
            elif isinstance(data, list):
                print(f"   JSON list with {len(data)} items")
            else:
                print("   JSON response (other type)")
        except ValueError:
            print("   Text response (first 300 chars):")
            print("   " + r.text[:300].replace("\n", " ") + " ...")
    
    except requests.exceptions.RequestException as e:
        print(f"❌ Request failed: {e}")
```

#### Interface consistency
*Involves checking naming conventions, field types and data formats, error messages, cross-platform support (language neutral data representation, no additional data transformations needed when consuming the data)*

In the case of an OGC SensorThings API implementation, checking the interface consistency means ensuring that your service follows the OGC standard and behaves predictably across resources, parameters, and responses.

```python
# 1. Check resource naming & structure 
# Ensure all entity sets are exposed (Things, Locations, Datastreams, Observations, ObservedProperties, Sensors, FeaturesOfInterest, etc.).
#   --> confirmed through API exploration

# Confirm case sensitivity: OGC STA uses PascalCase entity names (e.g., Datastreams, not datastreams).
#   --> confirmed through API exploration

# Check navigation links: $expand and $select should resolve consistently: 
def check_expand_consistency(BASE_URL, THING_ID):
    # 1. Direct navigation
    direct_resp = requests.get(f"{BASE_URL}/Things({THING_ID})/Datastreams")
    direct_resp.raise_for_status()
    direct = direct_resp.json().get('value', [])

    # 2. Expanded inside Thing
    expand_resp = requests.get(f"{BASE_URL}/Things({THING_ID})?$expand=Datastreams")
    expand_resp.raise_for_status()
    expanded = expand_resp.json().get('Datastreams', [])

    # 3. Compare IDs
    ids_direct = {d["@iot.id"] for d in direct}
    ids_expanded = {d["@iot.id"] for d in expanded}

    if ids_direct == ids_expanded:
        print("✅ $expand matches direct navigation.")
    else:
        print("❌ Mismatch in expanded vs direct navigation!")
        print("Direct IDs:  ", ids_direct)
        print("Expanded IDs:", ids_expanded)


def check_select_consistency(BASE_URL, THING_ID):
    # Request only name + description
    select_resp = requests.get(
        f"{BASE_URL}/Things({THING_ID})?$select=name,description"
    )
    select_resp.raise_for_status()
    selected = select_resp.json()

    # Mandatory OData fields: @iot.id, @iot.selfLink
    allowed_fields = {"name", "description", "@iot.id", "@iot.selfLink"}

    extra_fields = set(selected.keys()) - allowed_fields
    if not extra_fields:
        print("✅ $select returned only requested properties.")
    else:
        print("❌ $select returned extra properties:", extra_fields)


thing_id = 11966
check_expand_consistency(endpoint, thing_id)
check_select_consistency(endpoint, thing_id)
```

```python
# 2. Data type & schema consistency

# Check that IDs are consistent: @iot.id must be unique and stable across requests.
def check_unique_ids(BASE_URL):
    """Check @iot.id uniqueness across Things."""
    url = f"{BASE_URL}/Things?$top=1000"  # adjust or implement paging for large datasets
    r = requests.get(url)
    r.raise_for_status()
    things = r.json().get("value", [])
    ids = [t["@iot.id"] for t in things if "@iot.id" in t]
    if len(ids) == len(set(ids)):
        print("✅ All @iot.id values are unique for (top 1000) Things")
    else:
        print("❌ Duplicate @iot.id values detected in (top 1000) Things!")


# Validate GeoJSON compliance for Locations and FeaturesOfInterest
# GeoJSON schema for Feature 
geojson_schema = {
    "type": "object",
    "required": ["type", "geometry", "properties"],
    "properties": {
        "type": {"const": "Feature"},
        "geometry": {
            "type": "object",
            "required": ["type", "coordinates"],
            "properties": {
                "type": {
                    "enum": [
                        "Point",
                        "LineString",
                        "Polygon",
                        "MultiPoint",
                        "MultiLineString",
                        "MultiPolygon",
                        "GeometryCollection"
                    ]
                },
                "coordinates": {"type": "array"}
            }
        },
        "properties": {"type": "object"}
    }
}


def check_geojson(BASE_URL, resource="Locations"):
    """Check GeoJSON compliance for Locations or FeaturesOfInterest."""
    url = f"{BASE_URL}/{resource}?$top=100"
    r = requests.get(url)
    r.raise_for_status()
    entities = r.json().get("value", [])
    for e in entities:
        if "location" in e:  # Locations
            candidate = e["location"]
        elif "feature" in e:  # FeaturesOfInterest
            candidate = e["feature"]
        else:
            continue
        try:
            validate(candidate, geojson_schema)
        except ValidationError as err:
            print(f"❌ {resource}({e.get('@iot.id')}) invalid GeoJSON: {err.message}")
        else:
            pass
    print(f"✅ {resource} passed GeoJSON validation (checked {len(entities)} entities)")


# Verify time fields (phenomenonTime, resultTime) follow ISO 8601
def check_time_fields(BASE_URL):
    """Verify time fields are valid ISO8601 in Observations."""
    url = f"{BASE_URL}/Observations?$top=100"
    r = requests.get(url)
    r.raise_for_status()
    obs = r.json().get("value", [])
    for o in obs:
        for field in ["phenomenonTime", "resultTime"]:
            if field in o:
                try:
                    isoparse(o[field])
                except Exception:
                    print(f"❌ Observation({o.get('@iot.id')}) has invalid {field}: {o[field]}")
    print(f"✅ Checked {len(obs)} Observations for ISO8601 time fields")


check_unique_ids(endpoint)
check_geojson(endpoint)
check_time_fields(endpoint)
```

```python
# 3. Cross-resource consistency
## 3.1 forward vs. reverse nagivation 
def check_navigation_consistency(BASE_URL, thing_id):
    print("Checking navigation consistency for Thing:", thing_id)
    # Forward navigation: Thing → Datastreams
    ds_url = f"{BASE_URL}/Things({thing_id})/Datastreams"
    print(ds_url)
    r = requests.get(ds_url)
    
    r.raise_for_status()
    ds_from_thing = r.json().get("value", [])

    for ds in ds_from_thing:
        ds_id = ds["@iot.id"]
        resp = requests.get(f"{BASE_URL}/Datastreams({ds_id})/Things")

        if resp.status_code != 200:
            print(f"⚠️ No JSON returned for Datastream({ds_id}) → Thing (status {resp.status_code}) | ❌ Datastream({ds_id}) does not point back to Thing({thing_id})")
            continue

        try:
            thing = resp.json()
        except ValueError:
            print(f"⚠️ Datastream({ds_id}) → Thing did not return JSON")
            continue

        if thing.get("@iot.id") == thing_id:
            print(f"✅ Datastream({ds_id}) ↔ Thing({thing_id}) consistent")
        else:
            print(f"❌ Datastream({ds_id}) does not point back to Thing({thing_id})")


check_navigation_consistency(endpoint, 12504)
```

```python
## 3.2 Datastream ↔ Observations consistency
def check_observation_consistency(BASE_URL, datastream_id=1):
    print("Checking navigation consistency for datastream:", datastream_id)
    
    # Datastream → Observations
    print(f"{BASE_URL}/Datastreams({datastream_id})/Observations")

    obs_from_ds = requests.get(f"{BASE_URL}/Datastreams({datastream_id})/Observations").json()["value"]

    # Observation → Datastream
    for obs in obs_from_ds:
        obs_id = obs["@iot.id"]
        ds_url = f"{BASE_URL}/Observations({obs_id})/Datastreams"
        ds =  requests.get(ds_url).json()["value"][0]
        if ds["@iot.id"] != datastream_id:
            print(f"❌ Observation({obs_id}) does not point back to Datastream({datastream_id})")
        else:
            print(f"✅ Observation({obs_id}) ↔ Datastream({datastream_id}) consistent")


check_observation_consistency(endpoint, "urn:observation:template:00136-35")
```

```python
# 4. Response & Error handling

# Verify HTTP status codes: 
#                           200 for success
#                           400 for invalid query
#                           404 for missing entity
#                           500 only for real server errors)

# Check that error messages follow OData error format (JSON with error.code, error.message)
def test_invalid_expand_select(BASE_URL):
    # 1. Invalid $expand
    r1 = requests.get(f"{BASE_URL}/Things(1)?$expand=InvalidNav")
    if r1.status_code == 400:
        try:
            err = r1.json()
            if "error" in err and "message" in err["error"]:
                print("✅ Invalid $expand returned proper OData error:", err["error"]["message"])
            else:
                print("⚠️ Invalid $expand returned 400 but JSON not in OData format")
        except ValueError:
            print("❌ Invalid $expand did not return JSON")
    else:
        print(f"❌ Invalid $expand returned status {r1.status_code} instead of 400")
        print(r1.content)

    # 2. Invalid $select
    r2 = requests.get(f"{BASE_URL}/Things(1)?$select=NonExistentField")
    if r2.status_code == 400:
        try:
            err = r2.json()
            if "error" in err and "message" in err["error"]:
                print("✅ Invalid $select returned proper OData error:", err["error"]["message"])
            else:
                print("⚠️ Invalid $select returned 400 but JSON not in OData format")
        except ValueError:
            print("❌ Invalid $select did not return JSON")
    else:
        print(f"❌ Invalid $select returned status {r2.status_code} instead of 400")
        print(r2.content)


test_invalid_expand_select(endpoint)
```

#### Versioning & Backwards compatilibility 
*Involves checking whether the API has a clear versioning strategy and maintains support for older clients when changes to the schema occur*

- version directly included in endpoint url: https://sextant.ifremer.fr/examind/WS/sts/coriolis/v1.1
- OGC SensorThings API protocol specifies versioning and backward-compatibility (see: https://www.ogc.org/standards/sensorthings/ )
- STAplus extension (v1.0) is explicitely designed to be fully backwards-compatible with both SensorThings API versions 1.0 and 1.1

Backward-Compatible Evolution & Versioning
- Part 1 Sensing: Versions 1.0 and 1.1—the latter adds more flexible properties (and in Observations, parameters) fields for richer metadata. 
- Part 2 – Tasking Core: Version 1.0

```python
from urllib.parse import urlparse
def detect_version_from_url(url):
    """
    Detects SensorThings API version from the URL.
    Returns '1.0', '1.1', or 'Unknown'.
    """
    # Normalize URL
    path = urlparse(url).path.lower()
    
    if "/v1.0" in path:
        return "1.0"
    elif "/v1.1" in path:
        return "1.1"
    else:
        return "Unknown"



def detect_sensorthings_version(base_url):
    """
    Detects if the SensorThings API endpoint is v1.0 or v1.1.
    Returns '1.0', '1.1', or 'Unknown'.
    """
    # Common version paths
    version_paths = ["v1.0", "v1.1"]
    
    for version in version_paths:
        url = f"{base_url.rstrip('/')}/{version}/$metadata"
        try:
            response = requests.get(url, timeout=5)
            if response.status_code == 200:
                text = response.text
                # Check for v1.1-specific fields in metadata
                if "properties" in text or "parameters" in text:
                    return "1.1"
                else:
                    return "1.0"
        except requests.RequestException:
            continue
    
    return "Unknown", response.status_code

# Example usage 
url_version = detect_version_from_url(endpoint)
print(f"Detected SensorThings API version from URL: {url_version}")

metadata_version = detect_sensorthings_version(endpoint)
print(f"Detected SensorThings API version from metadata URL: {metadata_version}")
```
