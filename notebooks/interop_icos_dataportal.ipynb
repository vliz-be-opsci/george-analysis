#### Accessible Data Formats
*Users can download and work with data easily (Offer CSV, JSON, XLSX, GeoJSON in addition to visual charts)*

#### Persistent, Shareable Links
*Supports collaboration and citations (Stable dataset URLs; query parameters preserved in sharable links)*

Findings:
- data records have a URL with a clear, short and human-readable structure 
- search parameters contained in fragment identifier - rearranging parameter order doesn't influence results
- no dependency on UI state alone (same content independent of browser)
- versioned resources identified with different PIDs and linked via metadata (version not in URL)
- long-term hosting commitment (stable and controlled domain)
- correct use of http status codes 
- not preview friendly (when shared on social media, teams, ... no preview is displayed)

### Semantic interoperability

#### Clear Dataset Titles & Descriptions
*Prevents misinterpretation of content (Avoid acronyms in titles; describe scope and source in plain language)*

Findings:
- different version of meatadata record have different amounts of metadata information 
    - HTML has clear metadata, decent amount, with links
    - json has more metadata that HTML
        - more information available 
        - key of key:value-pairs follow schema.org term to certain extent, this facilitates conversion to jsonld
    - RDF/Turtle has far less metadata than HTML, while RDF allows for rich metadata
        - not all info described in triples
        - contains terms from self defined ontology, rather than community standard ontologies such as prov-o, dcat, ...
    - RDF/XML also has (very) little metadata 
    - XML (ISO 19115) has (little) metadata describing geographic information
- titles contain acronyms
- descriptions can be short

#### Standard Terminology in UI
*Improves findability and understanding (Use ISO, INSPIRE, or other domain vocabularies for categories and tags)*

- use of standard terms from domain vocabularies for data type, thematic centre, affiliation, station, responsible organization, .... 
    - domain vocabularies with own namespace
    - ideally, the standard terms also have rdf, json, ttl, jsonld serializations (now only html appears available)
- use of standard terms for license (Creative Commons, ICOS license, ...)
- date are presented in ISO 8601
- additionally standard terms are provided with an href ~ directing to a (HTML) desciption of the terms  

#### Consistent Units and Formats
*Avoids confusion when comparing datasets (Always display units with values (“Population: 2.3 million” not “Population: 2.3”))*

- sometimes units are present, when 'previewable variables' are available, (then data file itself also contains/documents the units)
- sometimes no variables, nor units are avaible, (then data file itself doens't contain/document units) 

#### Data Dictionaries Accessible from Browser
*Users can quickly interpret columns or codes (Tooltip explanations, downloadable glossary)*

Findings:
- search interface contains pop-up that explain meaning of search parameter values
- standard terms with href allow to retrieve (some kind of) explanation of the term
- when downloading a data file, information available on the license, how to cite, PIDS, ....
- meaning of acronyms not always available, e.g. when in title
- when "previewable variables" are available, variables name or value type could be better described for correct interpretation (e.g. use standard term to refer to 'P_sal [psu]')
