#### Conformance to protocols, content-type and content negotiation

````python
response = requests.get(endpoint)
server_headers = response.headers

tech_results = {
    # General protocol and status checks
    "General protocol and status checks" : 
        {
        "HTTP/HTTPS Protocol": endpoint.startswith("http"),
        "Status Code 200 OK": response.status_code == 200,
        },
    # Content negotiation
    "Content negotiation" : 
        {
        "MIME Type Present": "Content-Type" in server_headers,
        "MIME Type": server_headers.get("Content-Type"),
        "Content Negotiation Support": "Vary" in server_headers or "Accept" in server_headers,
        "Vary": server_headers.get("Vary"),                         # response may change depending on 'Accept-Encoding'
        "Accept-Ranges": server_headers.get("Accept-Ranges"),       # support partial content requests (e.g., 'bytes')
        },
    # Content negotiation related headers
    "Content negotiation related headers" : 
        { 
        "Content-Length": server_headers.get("Content-Length"),             # total size of the file, if specified
        "Content-Disposition": server_headers.get("Content-Disposition"),   # tells if file is inline or attachment
        "Content-Encoding": server_headers.get("Content-Encoding"),         # compression format (e.g., gzip)
        "Accept": server_headers.get("Accept"),                             # what media types the client accepts (rare in response)
        "Accept-Encoding": server_headers.get("Accept-Encoding"),           # what compression formats the client accepts (rare in response)
        },
    # Caching-related headers
    "Caching-related headers" : 
        {
        "Cache-Control": server_headers.get("Cache-Control"),       # caching policy (e.g., no-cache, max-age)
        "ETag": server_headers.get("ETag"),                         # version identifier for the file, useful for caching and validation
        }
}

# Output the results
tech_results
````

Findings:  
**CSV / Argo text files:**
- These files are inherently flat and human-readable, making them easy to open and inspect.
- However, automated processing is challenging because:
    - Column names are not always consistent across files.
    - Many cell values are codes or numeric identifiers (e.g., DAC_CODE, PLATFORM_CODE, OLD_QC, NEW_QC) whose meaning is not explicitly documented.
    - Some columns contain mixed data types (strings, numbers, URLs), which complicates parsing.
    - Free-form text entries and irregular formatting may require additional cleaning before analysis.
- URLs included in the files could provide extra context or references, but automated scripts must handle them separately and may fail if links are broken.

**NetCDF files:**
- These files are highly structured, with clearly defined dimensions, variables, and attributes.
- Libraries such as netCDF4 or xarray provide direct programmatic access, enabling reliable extraction of data arrays.
- Multi-dimensional variables (e.g., temperature profiles over depth and time) require careful alignment of dimensions to ensure proper interpretation.
- Overall, NetCDF files are highly machine-readable, allowing automated pipelines to work efficiently with minimal preprocessing compared to CSV or text files.

**HTML files:**
- HTML is designed for human browsing, so machine access is less straightforward.
- Programmatic extraction requires parsing the HTML structure (tables, lists, nested tags) using tools like BeautifulSoup.
- Column labels, codes, and metadata may be embedded inconsistently within free text or tables, making automated processing more difficult.
- Additional preprocessing is required to extract, clean, and normalize the data for use in downstream applications.

##### 2. NetCDF files

````python
# function to download .nc file
def download_nc_file(url):
    os.mkdir("files_euroargo_https") if not os.path.exists("files_euroargo_https") else None
    filename = url.split("/")[-1]
    file_path = os.path.join("files_euroargo_https", filename)

    #print(f"Downloading {url} to {file_path}...")
    response = requests.get(url)
    response.raise_for_status()           # Raise an error if download failed

    with open(file_path, 'wb') as f:
        f.write(response.content)
    #print("Download complete.")


#function to print metadata from a netCDF file
def nc_metadata(file_path):
    with Dataset(file_path, 'r') as nc_file:
        print("="*60)
        print(f"Metadata for: {file_path}")
        print("="*60)

        # Global Attributes
        print("\n>>> Global Attributes:")
        for attr in nc_file.ncattrs():
            print(f"  - {attr}: {getattr(nc_file, attr)}")

        # Dimensions
        print("\n>>> Dimensions:")
        for name, dim in nc_file.dimensions.items():
            print(f"  - {name}: size = {len(dim)}")

        # Variables
        print("\n>>> Variables:")
        for name, var in nc_file.variables.items():
            dims = ", ".join(var.dimensions)
            print(f"  - {name} ({dims}) -> shape = {var.shape}")
            
        print("="*60 + "\n")
````

````python
# Example of random profile nc file
prof_url = "https://data-argo.ifremer.fr/geo/atlantic_ocean/1997/07/19970728_prof.nc"
# Downloading file
download_nc_file(prof_url)
# Getting the  metadata
nc_metadata("files_euroargo_https/19970728_prof.nc")
````

Findings:  

**CSV / Argo text files:**
- Metadata is present but often incomplete or not standardized.
- Many columns contain codes or numeric identifiers (e.g., DAC_CODE, PLATFORM_CODE, OLD_QC, NEW_QC) that likely reference standard terms or parameters, but their meaning is not explicitly documented and no external mapping is provided.
- Relationships between observations, parameters, or measurement cycles are often implicit, making it harder for systems to understand the full context of the data.
- Free-form descriptions or notes may vary across files, further limiting semantic consistency.

**NetCDF files:**
- Variables and attributes are generally well-documented and standardized, often following Argo conventions for variable names and units.
- Global attributes provide provenance, source information, and context, which is critical for semantic interoperability.
- Even with good standardization, full semantic interoperability requires consistent use of controlled vocabularies and external references; deviations or non-standard terms still require mapping to a common schema.
- Overall, NetCDF files provide strong support for semantic interoperability, making it easier to integrate and interpret data across datasets and systems.

**HTML files:**
- Metadata is present within tables, lists, or free-form text, but its meaning is not standardized, and key relationships may be implied rather than explicit.
- URLs and embedded links can provide additional context or references, but automated interpretation requires additional mapping and processing.
- Semantic interoperability is limited because systems cannot directly infer the meaning of codes, labels, or values without human guidance or additional metadata mapping.

##### 2. NetCDF files

````python
# Check for CF Conventions declaration
def check_cf_conventions(nc_file):    
    if 'Conventions' in nc_file.ncattrs():
        print(f"Conventions: {nc_file.Conventions}")
        if "CF" in nc_file.Conventions:
            print("✅ CF Conventions are declared.")
        else:
            print("⚠️ Conventions declared, but not CF.")
    else:
        print("❌ No 'Conventions' attribute found.")

    print("\nCF Standard Names in Variables:")
    found = False
    for name, var in nc_file.variables.items():
        if hasattr(var, 'standard_name'):
            print(f"{name}: standard_name = {var.standard_name}")
            found = True

    if not found:
        print("⚠️ No 'standard_name' attributes found in any variables.")

    for name, var in nc_file.variables.items():
        if hasattr(var, 'units'):
            print(f"{name}: units = {var.units}")
    
    return None
````

##### 4. HTML files 

````python
def check_standard_vocabularies(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    findings = {
        "meta_tags": [],
        "rdfa_properties": [],
        "schema_org_items": [],
        "json_ld": []
    }

    # Check <meta> tags for Dublin Core or schema-related names
    for meta in soup.find_all("meta"):
        name = meta.get("name", "")
        if name and ("dc" in name.lower() or "dublin" in name.lower() or "schema" in name.lower()):
            findings["meta_tags"].append(str(meta))

    # Check RDFa (property, typeof, vocab attributes)
    for tag in soup.find_all(attrs={"property": True}):
        findings["rdfa_properties"].append(str(tag))

    # Check Schema.org via itemscope/itemtype/itemprop
    for tag in soup.find_all(attrs={"itemscope": True, "itemtype": True}):
        findings["schema_org_items"].append(str(tag))

    # Check JSON-LD blocks
    for script in soup.find_all("script", type="application/ld+json"):
        try:
            findings["json_ld"].append(json.loads(script.string))
        except Exception:
            findings["json_ld"].append(script.string)

    return findings
````

````python
html_url = "https://data-argo.ifremer.fr/aux/aoml/1901608/1901608_tech_aux.html"
results = check_standard_vocabularies(html_url)

print("---- Meta Tags (Dublin Core / Schema) ----")
for m in results["meta_tags"]:
    print(m)

print("\n---- RDFa Properties ----")
for r in results["rdfa_properties"]:
    print(r)

print("\n---- Schema.org Items ----")
for s in results["schema_org_items"]:
    print(s)

print("\n---- JSON-LD ----")
for j in results["json_ld"]:
    print(j)
````

Findings on use of standard vocabularies and ontologies in various file types:  
**Text files**
- No explicit reference to controlled vocabularies or ontologies.
- Codes are used (e.g., profiler_type, institution) but their meaning is not documented or mapped to external standards.
- Column semantics are unclear in places (e.g., file), which reduces interpretability.
- Some concepts such as project and institution could, in principle, be unambiguously linked to external identifiers or URIs, but this is not implemented.

**NetCDF files**
- CF (Climate and Forecast) conventions are explicitly declared, ensuring standardization of variable names and units.
- Use of recognized vocabularies and ontologies is evident (see [interop_euroargo_erddap_service](./interop_euroargo_erddap_service.ipynb) for detailed exploration).
- This structured adherence to standards makes NetCDF files the most semantically interoperable of the formats assessed.

**CSV files**
- Date formats are present but not ISO 8601 compliant (YYYY-MM-DD), limiting interoperability.
- No geographic metadata is provided in the assessed files, preventing conformance with ISO 19115 standards for geospatial information.
- No explicit declaration of CF conventions, although the PARAMETER column in ar_scoop2_KO_20141107123100.csv suggests partial use of standard names.
- Heavy reliance on codes (e.g., DAC_CODE, QC flags) without external mapping, making interpretation difficult.
- The PI column shows inconsistent naming conventions (uppercase vs. capitalized).
- Overall, CSV files show very limited use of standard vocabularies or ontologies.

**HTML files**
- No evidence of controlled vocabularies, ontologies, or machine-readable metadata standards (e.g., Dublin Core, Schema.org, RDFa, JSON-LD) was found in the assessed pages.

**Notes on granularity**  

Data granularity refers to the level of detail at which data and metadata are made available.
- Granularity of metadata:  
Metadata is typically provided at the file level.  
It can either be accessed through separate metadata files (retrieved by navigating directory structures) or found embedded directly within the data files themselves.

- Granularity of data:  
Data is also provided at the file level, with each file containing the actual ocean measurements or observations.  
Accessing these files may require navigating directory structures or constructing download links by combining a base URL with relative paths. In some cases, an additional step is necessary, where the base URL must be combined with paths listed inside the metadata files to obtain the actual NetCDF data.
