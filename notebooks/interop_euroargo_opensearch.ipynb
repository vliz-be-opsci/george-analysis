### Exploring the OpenSearch service

Documentation is available at: https://opensearch.ifremer.fr/html/docs/0.1/getting-started/introduction/

dynamic openapi documentation available at https://opensearch.ifremer.fr/openapi#
- /health ~ metadata on health status of service; no access
- /collections ~ overview of Dataset IDs (coming from 2 datasources: ARGO In-Situ Datasource (1 datasetID) and CERSAT Satellite Datasource (all other datasetIDs))
- /collections.atom; parameter=DatasetID ~ metadata about Dataset record  
- /granules.description ~ description of parameters available to query data itself 
- /granules.atom; parameter=DatasetId, other optional parameters ~ to get list of granules with metadata, optionally filtered to specified parameters  
(if no optional parameters are specified, returns metadata of the first 1000 datafiles in the datasetID/collection)  
(url in the <id> property returns following {"detail":[{"loc":["query","datasetId"],"msg":"field required","type":"value_error.missing"}]} )  
(**CERSAT Satellite Datasource**: href in <link> property returns relative url that doesn't allow direct download (documentation indicates that files are on private FTP and HTTPS servers))  
(**ARGO In-Situ Datasource**: href in <link> property returns absolute url that allows direct download of metadata or data (via FTP or HTTPS))  


granules ~~ NetCDF files from a wide range of datasets, with search criteria such as bounding box, timespan, and some more properties depending on the datasource.
