# Interoperability Assessment by Service Type

#### Interoperability in the Context of Data Services:
In the data services context, interoperability refers to the ability of different data systems and platforms to discover, access, exchange, integrate, and use data in a coherent and consistent way—regardless of differences in data format, structure, origin, or software environment.

**Goal**: Enable seamless data integration, reusability, and automation across systems—whether within a single organization or across institutions or sectors (e.g., scientific research, government open data, commercial analytics).

#### Assessment

Starting from the [overview of dataservices](Overview_ServiceEndpoints.ipynb), data services were categorized into different types based on their characteristics and assessed for their interoperability.  
In this assessment, we considered several key aspects: 
- Access protocols and authentication mechanisms (e.g., HTTPS, OAuth2, API keys)

- Standardized data formats (e.g., CSV, JSON, XML, NetCDF, Parquet)

- Consistent metadata schemas (e.g., Dublin Core, DCAT, ISO 19115)

- Standard APIs for querying and retrieving data (e.g., OGC APIs, RESTful APIs)

- Semantic interoperability, ensuring that the meaning of the data is preserved across systems through shared vocabularies or ontologies

- Syntactic interoperability, meaning the data structure can be read and parsed by other systems


### 1. HTTPS Server

- https://data-argo.ifremer.fr/  

**Findings**:  
| Dimension              | Evaluation |
|------------------------|------------|
| Protocol               | HTTP (read-only, no API logic) |
| File discovery         | Manual or directory scraping |
| Metadata exposure      | Missing or embedded in files |
| Format                 | text files and standard formats like NetCDF and CSV |
| Integration effort     | High — often no structured metadata, requires parsing |
| Interoperability rating| ⭐ Low — poor automation support |

**Comments**  
- Serves primarily to store and share files, in formats like text, CSV or NetCDF ~ Suitable for archival or download, but not interoperable  
- Accessing & navigating data requires additional steps and knowledge of file structure 
- While the data is accessible, the lack of structured, queryable interfaces means that other systems must download and process the files before they can use the data

### 2. FTP Server
Traditional file-hosting protocol used for bulk data transfer, often seen in legacy systems and long-standing data repositories.

**Endpoints of this type**:
- ftp://ftp.ifremer.fr/ifremer/glider/v2 

**Findings**:
| Dimension               | Evaluation                                                              |
| ----------------------- | ----------------------------------------------------------------------- |
| Protocol                | FTP (read-only, no API logic)                                           |
| File discovery          | Manual navigation or scripted directory crawling                        |
| Metadata exposure       | Typically missing or provided as separate README or index files         |
| Format                  | Structured formats like NetCDF, HDF, CSV; often organized by directory  |
| Integration effort      | High — requires file parsing and manual or scripted metadata extraction |
| Interoperability rating | ⭐ Low — not designed for modern data interoperability                   |

**Comments**
- FTP servers function primarily as bulk data archives, offering file-based access without standardized query mechanisms
- Users must know the directory layout and file naming conventions to locate relevant data
- Metadata, if present, is often unstructured or provided in separate documentation files
- Integration into automated workflows requires custom scripting for navigation, downloading, and parsing
- While reliable for downloading large datasets, FTP lacks support for real-time data discovery or semantic interoperability

**Endpoints of this type**:

**Findings**:
| Dimension               | Evaluation                                                                       |
| ----------------------- | -------------------------------------------------------------------------------- |
| Protocol                | HTTPS (Amazon S3 API)                                                            |
| File discovery          | Supported via AWS CLI, SDKs, or GUI listings                                     |
| Metadata exposure       | Partial — often embedded in object names or provided via separate metadata files |
| Format                  | Standard formats like NetCDF, JSON, CSV                                          |
| Integration effort      | Medium — metadata may require interpretation; access via S3 tools or API         |
| Interoperability rating | ⭐⭐ Medium — good access protocol, but discovery and metadata vary                |

**Comments**

- Metadata may be implicit (e.g., in filenames or paths) or provided in separate files; semantic structure is not enforced
- Access can be automated using AWS CLI, SDKs, or compatible tools, but querying within the data requires downloading or external indexing
- Well-suited for bulk access and integration into data pipelines, but not inherently interoperable without supporting metadata services

### 4. Open Search 
Search-oriented interface for discovering metadata and datasets via standardized query parameters, often used in geoscience and Earth observation data portals.

**Endpoints of this type**:
- https://opensearch.ifremer.fr/html

**Findings**:
| Dimension               | Evaluation                                                                        |
| ----------------------- | --------------------------------------------------------------------------------- |
| Protocol                | HTTPS with OpenSearch Description Document (OSDD)                                 |
| File discovery          | Queryable via standard OpenSearch parameters (e.g., time, bounding box, keywords) |
| Metadata exposure       | Present — provided in structured XML or Atom/GeoRSS responses                     |
| Format                  | Pointers to downloadable resources (e.g., NetCDF, CSV, ZIP)                       |
| Integration effort      | Low to Medium — standardized interface, but responses may need parsing            |
| Interoperability rating | ⭐⭐⭐ Medium to High — supports structured discovery but limited data interaction   |

**Comments**
- OpenSearch enables metadata-based search and filtering, making dataset discovery easier than with raw file servers
- The service returns structured metadata and links to downloadable data, often in standard formats
- While it doesn't offer direct data interaction (e.g., subsetting or querying within datasets), it improves interoperability by exposing structured metadata and standardized search endpoints
- Parsing responses may require XML or feed handling, but the query interface is documented and machine-friendly
- Candidate for integration into federated search portals or harvesting systems

### 5. ERDDAP Server
Web service that provides a standardized interface for discovering, querying, visualizing, and downloading scientific datasets—particularly time-series and gridded data.

**Endpoints of this type**:
- https://erddap.ifremer.fr/erddap/tabledap/ArgoFloats-synthetic-BGC.graph
- https://erddap.ifremer.fr/erddap/tabledap/OceanGlidersGDACTrajectories.html
- https://erddap.emso.eu/erddap


**Findings**:
| Dimension               | Evaluation                                                                                                  |
| ----------------------- | ----------------------------------------------------------------------------------------------------------- |
| Protocol                | HTTPS with RESTful endpoints and table/griddap APIs                                                         |
| File discovery          | Searchable catalog with filters and keyword queries                                                         |
| Metadata exposure       | Rich — standardized metadata (ISO 19115, FGDC, etc.) exposed in multiple formats (e.g., .html, .xml, .json) |
| Format                  | Downloadable in various standard formats (CSV, NetCDF, JSON, GeoJSON, .mat, .nc)                            |
| Integration effort      | Low — data and metadata access is well-documented and scriptable                                            |
| Interoperability rating | ⭐⭐⭐⭐ High — designed for interoperability and automated use                                                 |


**Comments**:
- ERDDAP (Environmental Research Division's Data Access Program) serves primarily provide data in NetCDF format, with support for multiple alternative output formats.
- An analysis of property alignment across different ERDDAP servers reveals partial consistency in naming, but many properties remain unique to individual servers.
- The limited use of persistent identifiers and standardized vocabularies contributes to semantic ambiguity, which hinders full interoperability.
- While ERDDAP offers greater interoperability than basic file servers—thanks to its standardized APIs and flexible data outputs—the variation in property definitions and inconsistent semantics across instances remain significant barriers to seamless integration.

### 6. OGC SensorThings API
A RESTful, standards-based API for accessing sensor observations and metadata, designed to support real-time IoT and environmental monitoring systems.

**Endpoints of this type**:
- https://sextant.ifremer.fr/examind/WS/sts/coriolis/v1.1

**Findings**:
| Dimension               | Evaluation                                                                                         |
| ----------------------- | -------------------------------------------------------------------------------------------------- |
| Protocol                | HTTPS with RESTful OGC SensorThings standard                                                       |
| File discovery          | Queryable via structured endpoints (e.g., `/Things`, `/Datastreams`, `/Observations`)              |
| Metadata exposure       | Rich — entities such as Sensors, ObservedProperties, and FeaturesOfInterest are explicitly modeled |
| Format                  | JSON (OData-compliant responses), suitable for automated processing                                |
| Integration effort      | Low — standardized data model and API facilitate direct integration                                |
| Interoperability rating | ⭐⭐⭐⭐ High — strong syntactic and semantic structure supports automation and system integration     |

**Comments**:
- Delivers data in a structured JSON format, which is widely supported and easy to consume across diverse platforms and programming languages, facilitating broad usability.
- The API design and data models follow well-established standards such as the SensorThings specification, which simplifies integration into other applications, mobile apps, and larger systems.
- Although URLs and some standard terms are used within the data model to reference entities, limited use of persistent identifiers and controlled vocabularies creates semantic ambiguity, which hinders interoperability between different implementations.
- Overall, the SensorThings API offers decent syntactic and semantic interoperability through its standardized interfaces and rich metadata model, making it suitable for automated workflows, real-time data access, and federated sensor networks. However, semantic inconsistencies due to variations in how properties and units are defined across implementations require additional harmonization efforts.

### 7. SPARQL Endpoint
A query interface that allows retrieval and manipulation of data stored in RDF format using the SPARQL query language, commonly used in linked data and semantic web applications.

**Endpoints of this type**:
- https://sparql.ifremer.fr/argo/query
- https://meta.icos-cp.eu/sparqlclient/?type=CSV 

**Findings**:
| Dimension               | Evaluation                                                                                |
| ----------------------- | ----------------------------------------------------------------------------------------- |
| Protocol                | HTTPS with SPARQL query interface                                                         |
| File discovery          | Query-based discovery using SPARQL queries                                                |
| Metadata exposure       | Rich — metadata and data are exposed as linked RDF triples                                |
| Format                  | RDF serialization formats (e.g., JSON-LD, XML, Turtle)                                    |
| Integration effort      | Medium — requires knowledge of SPARQL and RDF data models                                 |
| Interoperability rating | ⭐⭐⭐⭐ High — strong semantic interoperability via standardized vocabularies and ontologies |

**Comments**:
- A SPARQL endpoint provides a standardized way to query RDF data using the SPARQL query language. RDF serves as a universal model for data interchange on the web, and the use of SPARQL enables highly interoperable data sharing across diverse systems, especially within the linked data and Semantic Web ecosystems.
- SPARQL endpoints excel in scenarios requiring integration of data from multiple heterogeneous sources, offering powerful and flexible query capabilities over rich semantic relationships.
- They expose detailed metadata and data as linked RDF triples, facilitating deep semantic connections and interoperability through shared vocabularies and ontologies.
- While they provide high semantic interoperability, effective use requires familiarity with SPARQL and RDF concepts, which can be a learning curve for some users.
- Performance and usability may vary depending on the endpoint implementation and dataset size, but overall, SPARQL endpoints are vital tools for semantic data integration and advanced querying.

### 9. Web-Based Data Portal
Graphical user interface (GUI) that allows users to interactively browse, filter, and download datasets, often aggregating multiple data types with human-readable context.

**Endpoints of this type**:
- https://data.icos-cp.eu/portal/

**Findings**:
| Dimension               | Evaluation                                                                                         |
| ----------------------- | -------------------------------------------------------------------------------------------------- |
| Protocol                | HTTPS (GUI-driven with optional programmatic access)                                               |
| File discovery          | Interactive filtering, search tools, and catalog browsing                                          |
| Metadata exposure       | Rich — human-readable metadata via the interface, sometimes exportable in machine-readable formats |
| Format                  | Downloadable files (e.g., NetCDF, CSV, JSON); occasionally APIs for programmatic access            |
| Integration effort      | Medium — GUI is user-friendly, but automated access depends on availability of backend APIs        |
| Interoperability rating | ⭐⭐ Medium — designed for human users; API or metadata export options vary                          |

**Comments**:
- Intuitive interfaces for human users to explore and access diverse datasets, often across spatial and temporal dimensions
- Typically expose rich metadata (e.g., descriptions, units, provenance), though it is often designed for human readability rather than structured machine consumption
- allows downloads of data in standard formats (e.g., CSV, NetCDF) but consistent, well-documented programmatic access (via APIs or metadata catalogs) is not always available
- Interoperability is moderate: the lack of persistent identifiers, semantic tagging, or machine-readable vocabularies can hinder integration with external systems
- Can serve as valuable bridges between human usability and system-level automation when complemented with APIs or metadata harvesting services (e.g., CSW, OAI-PMH)
